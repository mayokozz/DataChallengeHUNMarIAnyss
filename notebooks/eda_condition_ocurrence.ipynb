#Introducción

**Descripción: **Las afecciones (*conditions*) se definen mediante conceptos (*concepts*) pertenecientes al dominio de afecciones, los cuales conforman una jerarquía compleja. Como resultado, una misma persona con una misma enfermedad puede tener múltiples registros de afecciones que pertenecen a una misma familia jerárquica. La mayoría de los registros de afecciones se derivan de códigos de diagnóstico; no obstante, los signos, síntomas y descripciones resumidas que se hayan registrado también contribuyen a esta tabla. Los diagnósticos descartados (*rule-out diagnoses*) no deberían incluirse en esta tabla; sin embargo, en la práctica, su naturaleza excluyente no siempre queda reflejada en los datos de origen, por lo que es necesario adoptar precauciones adicionales al identificar a las personas que efectivamente padecen la afección registrada. Se deben registrar todas las afecciones tal como aparecen en los datos de origen. Cualquier decisión relativa a la definición de diagnósticos o fenotipos debe gestionarse mediante especificaciones de cohortes, las cuales pueden almacenarse en la tabla COHORT. Aunque las afecciones abarcan un intervalo temporal con un inicio y un fin, por lo general se registran como eventos puntuales sin fecha de finalización. Esto se debe a dos razones: (i) en el momento del registro, se desconoce la duración y esta no se añade posteriormente; y (ii) las personas suelen dejar de interactuar con el sistema sanitario cuando se sienten mejor, lo que provoca que las afecciones resueltas no queden registradas de forma completa. La tabla CONDITION_ERA aborda esta limitación. Los antecedentes familiares y los diagnósticos previos («historial de...») no se incluyen en esta tabla, sino que figuran en la tabla OBSERVATION. Asimismo, los códigos empleados durante el proceso de diagnóstico —como aquellos que indican «posible...» o «descartar...»— no deben representarse aquí; en su lugar, si se requieren para realizar análisis, deben registrarse en la tabla OBSERVATION. No obstante, cabe señalar que esta información no siempre está disponible.

```python
import pandas as pd
```

```python
condition_ocurrence = pd.read_csv("../data/condition_occurrence_202611061452.csv")
```

```python
#Rows and columns values
rows, column = (condition_ocurrence.shape)
print(f"rows: {rows} \ncolumn {column}")

total_visit = len(condition_ocurrence)
print(f"total_visit: {total_visit}")
```

```python
#Null values
list_columns = condition_ocurrence.columns.to_list()

for i in list_columns:
    num_nulls = condition_ocurrence[i].isnull().sum()
    
    if num_nulls > 0:
        print(f"The column '{i}' HAS nulls ({num_nulls} in total)")
    else:
        print(f"The column '{i}' HAS NO nulls")
```
