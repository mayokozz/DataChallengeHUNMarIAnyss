#Introduction

**Description: **The VISIT_DETAIL table is an optional table used to represents details of each record in the parent VISIT_OCCURRENCE table. A good example of this would be the movement between units in a hospital during an inpatient stay or claim lines associated with a one insurance claim. For every record in the VISIT_OCCURRENCE table there may be 0 or more records in the VISIT_DETAIL table with a 1:n relationship where n may be 0. The VISIT_DETAIL table is structurally very similar to VISIT_OCCURRENCE table and belongs to the visit domain.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
visit_detail = pd.read_csv("../data/visit_detail_202611061450.csv")
visit_occurrence = pd.read_csv("../data/visit_occurrence_202611061448.csv")
```

```python
#Table dimensions
rows, column = (visit_detail.shape)
print(f"rows: {rows} \ncolumn {column}")

total_visit = len(visit_detail)
print(f"total_visit: {total_visit}")
```

```python
#Null values
null_amount = visit_detail.isnull().sum()
null_porcent = (null_amount/total_visit)*100

summary_null = pd.DataFrame({
    'Missing values': null_amount,
    'Porcent (%)': null_porcent
})

print(summary_null.sort_values(by='Missing values', ascending=False))
```

```python
#Duplicate records
duplicate_records = visit_detail.duplicated().sum
columnas_afectadas = list(visit_detail.columns)

summary_duplicated = {
    "total_duplicated": duplicate_records,
    "columns_in": columnas_afectadas,
}

print(summary_duplicated)

duplicate_id = visit_detail["visit_detail_id"].duplicated().sum
print(duplicate_id)
```

```python
#Most frequent values
frequent_values = visit_detail["visit_detail_concept_id"].nlargest(10)
print(frequent_values)
```

```python
#Provider id register 
provider_id_register = len(visit_detail["provider_id"])
print("provider_id_register: ", provider_id_register)
```

```python
#Visit with detail
visit_with_detail= visit_occurrence["visit_occurrence_id"].isin(visit_detail["visit_occurrence_id"]).sum()
print("Visit with detail: ", visit_with_detail)
```

```python
#Visits without details
visits_witout_detail = (
    ~visit_occurrence["visit_occurrence_id"].isin(
        visit_detail["visit_occurrence_id"]
    )
).sum()

print(f"Visits without detail: {visits_witout_detail}")
```

```python
#Details regarding visits
details_per_visit = visit_detail["visit_occurrence_id"].value_counts()
detail_distribution = details_per_visit.value_counts().sort_index()
print(detail_distribution)

#Maximun detail 
print(f"The higher maximun detail is: {details_per_visit.max()}")
```

### VISIT_DETAIL – Key Findings

1. `VISIT_DETAIL` has a **95.12% coverage rate** relative to `VISIT_OCCURRENCE`, indicating that most visits have at least one associated detail record.

2. A single visit can be associated with multiple detail records, showing a **one-to-many relationship** between `VISIT_OCCURRENCE` and `VISIT_DETAIL`.

3. The number of detail records per visit ranges from **1 to 34** among visits with associated records. Most visits have a relatively small number of details.

4. The duration of visit details cannot be calculated because the start and end timestamps are identical for all records.

5. `provider_id` cannot be used to analyze provider distribution because all its values are zero, making this field uninformative for provider-level analysis.
